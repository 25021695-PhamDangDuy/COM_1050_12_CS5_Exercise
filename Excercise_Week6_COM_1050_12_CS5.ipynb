## <font color="palered">**W6A1 [RemoveDuplicates]**</font>

Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]

```python
def remove_duplicates(lst) -> list:
  check_set = set()
  box = []
  for i in lst:
    if i not in check_set:
      check_set.add(i)
      box.append(i)
  return box
n = (input())
lst_n = map(int, n.split())
print(remove_duplicates(lst_n))
```

## <font color="palered">**W6A2 [CumulativeSum]**</font>

Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự

```python
def sum_of_list(lst) -> list:
  box = []
  total = 0
  for i in lst:
    total += i
    box.append(total)
  return box

n = (input())
lst_n = map(int, n.split())
print(sum_of_list(lst_n))
```

## <font color="palered">**W6A5 [CountPosNegZero]**</font>

Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': ...} (đếm số >0, <0, =0)

```python
list_int = list(map(int, input().split()))
def count_number(lst):
  dict_target = {
  'positives' : 0,
  'negatives' : 0,
  'zeros' : 0
  }
  for i in lst:
    if i > 0: dict_target['positives'] += 1
    elif i < 0: dict_target['negatives'] += 1
    else: dict_target['zeros'] += 1
  return dict_target
print(count_number(list_int))
```

## <font color="palered">**W6A9**</font>

Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key.

```python
def merge_dict(dict1, dict2) -> dict:
  result = {}
  box = sorted(set(dict1) | set(dict2))

  for i in box:
    result[i] = dict1.get(i, 0) + dict2.get(i, 0)
  return result

def input_to_dict(a) -> dict:
  result = {}
  a = a.replace("'", "").replace("{", "").replace("}", "").replace(" ", "")
  if a == "":
    return {}

  st = a.split(",")
  for i in st:
    key, value = i.split(":")
    result[key] = int(value)
  return result

dict_a = input()
dict_b = input()
print(merge_dict(input_to_dict(dict_a), input_to_dict(dict_b)))
```

Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
def key_func(lst):
  return lst[0]
def find_pattern(lst, k) -> list:
  box = []
  set_box = set()
  for j in range(len(lst) - 1):

    for i in range(j + 1,len(lst)):
      val_j = lst[j]
      val_i = lst[i]

      if val_j + val_i == k and (val_j, val_i) not in set_box:
        box.append((val_j, val_i))
        set_box.add((val_j, val_i))

  box.sort(key = key_func)
  return box

n = (input())
lst_n = list(map(int, n.split()))
k = int(input())
print(find_pattern(lst_n, k))
```

## <font color="palered">**W6A11 [Tách số chẵn và lẻ từ tuple]**</font>

Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.

```python
def split_tuple(tup) -> tuple:
  tup_odd = []
  tup_even = []
  for i in tup:
    if i % 2 == 0:
      tup_even.append(i)
    else:
      tup_odd.append(i)
  return tuple(tup_even), tuple(tup_odd)
x, y = split_tuple(tuple(map(int, input().split())))
print(x)
print(y)
```

## <font color="palered">**W6A12 [Tìm phần tử xuất hiện nhiều nhất]**</font>

Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó.

```python

def repeat_number(lst) ->int:
  #Tạo dict tính số lần lặp lại của số
  count_dict = {}
  for i in lst:
    count_dict[i] = count_dict.get(i, 0) + 1

  #Thuật toán loại trừ
  result_val = None
  max_val = 0
  for key, value in count_dict.items():
    if value > max_val or (value == max_val and result_val > key):
      max_val = value
      result_val = key
  return result_val


n = list(map(int, input().split()))
print(repeat_number(n))


```

## <font color="palered">**W6A14 [Giao của hai danh sách]**</font>

Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên. Hãy tìm các phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các phần tử chung đó, không trùng lặp.

```python
def merge_list(lst_a, lst_b) -> list:
  result_list = []
  set_a = set(lst_a)
  set_b = set(lst_b)
  if len(set_a) > len(set_b):
    for x in set_b:
      if x in set_a:
        result_list.append(x)
  else:
    for x in set_a:
      if x in set_b:
        result_list.append(x)
  return result_list

a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(merge_list(a,b))


```

## <font color="palered">**W6A18**</font>

Nhậ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột k (hàng k)

```python
def check_row(str_lst, n) -> list:
  lst = list(map(int, str_lst.split()))
  if len(lst) == n:
    return lst
  else:
    return f"Phải nhập đúng {n} cột"

def create_matrix(row, col):
  matrix = [ check_row( input(f"Nhập các số hàng {i + 1}: "), col) for i in range(row) ]
  return matrix

def sum_k(matrix, k, key):
  positive_mtx = k - 1
  if key == "row":
    return sum(matrix[positive_mtx])
  elif key == "col":
    return sum( x[positive_mtx] for x in matrix)
  else:
    return f"tham số phải là {"row"} hoặc {"coll"} "

n, m , k = map(int, input().split())
matrix = create_matrix(n, m)
print(f"Tổng hàng {k}: {sum_k( matrix, k, "row")}")
print(f"Tổng cột {k}: {sum_k( matrix, k, "col")}")
```

## <font color="palered">**W6A21**</font>

Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
def positive_str(str_a) -> dict:
  result = {}
  lst = str_a.split()
  for i in range(len(lst)):
    if lst[i] not in result:
      result[lst[i]] = []
    result[lst[i]].append(i)
  for x,y in result.items():
    result[x] = tuple(y)
  return result
n = input()
print(positive_str(n))
```
