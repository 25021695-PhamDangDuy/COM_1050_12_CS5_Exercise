## <font color="palered">**W11A3**</font>
[ValueError - HappyNumbers]
Nhập vào một số nguyên dương a, Kiểm tra xem
a có phải là số hạnh phúc không.
Một số nguyên dương n được gọi là số hạnh phúc
nếu ta thực hiện liên tiếp các bước sau:
- Tính tổng bình phương các chữ số của n.
- Lặp lại bước đó với kết quả vừa tính.
- Nếu cuối cùng kết quả là 1, thì n là số hạnh
phúc.
- Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1
(ví dụ như 4 → 16 → 37 → ...), thì n không phải là
số hạnh phúc.
Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback
để gỡ lỗi nhập liệu. (Dùng try ... except bắt lỗi
ValueError thêm set)

```python

def sum_of_square(n):
  result = 0
  while n > 0:
    result += (n % 10) ** 2
    n //= 10
  return result
def is_happy_number(n):
  seen = set()
  while n != 1:
    print(n)
    if n in seen:
      return False
    seen.add(n)
    n = sum_of_square(n)
  print(n)
  return True


try:
  a = input()
  if not a.isdigit():
    raise ValueError("Phải là số nguyên")
  a = int(a)
  if a < 0:
    raise ValueError("Phải là số dương")
  print("True" if is_happy_number(a) else "False")
except ValueError as e:
  print(e)
```

## <font color="palered">**W11A4**</font>
Sử dụng try except,kết hợp cùng else... finally.
Tính tổng 2 số, nếu nhập sai kiểu thì thông báo,
luôn luôn in "Kết thúc chương trình." sau khi chạy

```python
try:
  a,b = input().split()
  if not a.isdigit() or not b.isdigit():
    raise ValueError("Phải là số nguyên")
  a, b = int(a), int(b)
  summary = a + b
  print(summary)
except ValueError as e:
  print(e)
finally:
  print("Kết thúc chương trình")
```

## <font color="palered">**W11A5**</font>
Viết chương trình chuyển tất cả phần tử của một
list cho trước về int.
Nếu phần tử nào trong list không thể chuyển được
thì in cảnh báo và bỏ qua phần tử đó


```python
a = ["3","a","2","1","10", "x"]
def change_int(i,a):
  try:
      if not a[i].isdigit():
        raise ValueError(f"{a[i]} không chuyển đổi được")
      a[i] = int(a[i])
  except ValueError as e:
    print(e)
for i in range(len(a)):
  change_int(i,a)
print(a)

```

chuỗi đó thành số nguyên (Ví dụ nhập '123'
chuyển thành 123). In ra số nguyên sau khi
chuyển. Nếu không thể chuyển, in ra "Chuoi

```python
try:
  string = input()
  is_negative = False
  if string[0] == "-":
    string = string[1:]
    is_negative = True
  if not string.isdigit():
    raise ValueError("Chuoi khong hop le")
  string = int(string)
  print(string * -1 if is_negative else string)
except ValueError as e:
  print(e)
```

## <font color="palered">**W11A10**</font>
Viết chương trình nhập vào phần tử N (N > 0) và
một mảng gồm N số nguyên dương. Một mảng là
hợp lệ nếu nó gồm các số khác nhau. In ra tổng
các phần tử nếu mảng đó là hợp lệ. Nếu mảng
không hợp lệ, in ra "Mang khong hop le".

```python
try:
  n = int(input())
  arr = list(input().split())
  if len(arr) != n:
    raise ValueError("Mảng không hợp lệ")
  total = 0
  seen = set()
  for i in range(len(arr)):
    if not arr[i].isdigit() :
      raise ValueError("Mảng không hợp lệ")
    arr[i] = int(arr[i])
    if arr[i] in seen:
      raise ValueError("Mảng không hợp lệ")
    total += arr[i]
    seen.add(arr[i])
  print(total)
except ValueError as e:
  print(e)
```

## <font color="palered">**W11A11**</font>
Cho string S, kiểm tra xem S có phải một regex
hợp lệ hay không.
Đầu vào: Dòng đầu tiên chứa số nguyên T - số
lượng testcase. T dòng tiếp theo mỗi dòng chứa
một chuỗi S.
Output: In ra "Valid" hoặc "Invalid" cho mỗi
testcase.
Ví dụ:
2
.*\+
.*+
Sample Output:
True
False
Giải thích:
.*+ : lỗi "multiple repeat" - invalid

```python
import re

t = int(input())

for _ in range(t):
    s = input().strip()
    try:
        re.compile(s)
        print("True")
    except re.error:
        print("False")
```
