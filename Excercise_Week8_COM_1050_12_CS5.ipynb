[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và
chiều cao (kiểu số thực) với hai phương thức:tính
diện tích bề mặt (diện tích toàn phần) và thể tích
của hình

**Đầu vào**

Đâu vào từ bàn phím gồm một dòng duy nhất
chứa bán kính và chiều cao của hình trụ, cách
nhau bởi một dấu cách.

**Đầu ra**

In ra màn hình diện tích bề mặt và thể tích của
hình trụ (sử dụng lại những hàm vừa viết, π=3.
14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
from math import pi
class Cylinder:
  def __init__(self,r,h):
    self.radius = r
    self.height = h
  def area(self):
    return (2 * (self.radius **2) * pi) + (2 * self.radius * pi * self.height )
  def  v(self):
    return  (self.radius ** 2) * pi * self.height

r, h = map(int,input().split())
n = Cylinder(r, h)
print(n.area())
print(n.v())
```

[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với
hai phương thức kiểm tra xem một ngày tháng
nào đó có hợp lệ hay không và phương thức tính
ngày kế tiếp của ngày tháng năm nào đó nếu nó là
hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm
biểu diễn dưới dạng dd/mm/yyyy). Nếu ngày
tháng năm không tồn tại, in ra dòng chữ INVALID.
Nếu là hợp lệ in ra ngày kế tiếp của nó.

**Đầu vào**
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng,
năm dưới dạng dd/mm/yyyy
.

**Đầu ra**
In ra màn hình ngày kế tiếp của ngày vừa nhập
dưới dạng dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID
.

```python
class Date:
  def __init__(self,d,m,y):
    self.day = d
    self.month = m
    self.year = y


  def _typeYear(self):
    if (self.year % 400 == 0) or (self.year % 100 != 0 and self.year % 4 == 0):
      self.calendar = [31,29,31,30,31,30,31,31,30,31,30,31]
    else:
      self.calendar = [31,28,31,30,31,30,31,31,30,31,30,31]

  def __str__(self):
    return "What's year?"

  def is_true(self):
    if self.month < 1 or self.month > 12:
      return False
    if self.day < 1 or self.day > self.calendar[self.month - 1]:
      return False
    return True

  def next_day(self):
    self.day += 1
    if self.day > self.calendar[self.month - 1]:
      self.day = 1
      self.month += 1
      if self.month > 12:
        self.month = 1
        self.year += 1
d,m,y = map(int,input().split("/"))
n = Date(d,m,y)
n._typeYear()
if n.is_true():
  n.next_day()
  print(f"Next day: {n.day}/{n.month}/{n.year}")
else:
  print("INVALID")
```

## <font color="palered">**W8A3**</font>
[Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng
các phương thức kiểm tra xem có phải gốc tọa dộ
không, có nằm trên trục tung không, có nằm trên
trục hoành không, tính khoảng cách từ điểm đó tới
một điểm khác

**Đầu vào:**
Nhập vào tọa độ một điểm M bất kỳ

**Đầu ra:**
Thông báo M có phải gốc tọa độ không hoặc có
nằm trên trục tung hoặc trục hoành không, khoảng
cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ
số phần thập phân)

```python
import math
class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y
  def is_O(self):
    return True if (self.x == 0 and self.y == 0) else False
  def is_Ox(self):
    return True if (self.y == 0) else False
  def is_Oy(self):
    return True if (self.x == 0) else False
  def distance(self,other):
    return math.sqrt((self.x ** 2 - other.x ** 2) + (self.y ** 2 - other.y ** 2))

x,y = map(int,input("Nhập M(x,y): ").split())
M = Point(x,y)
O = Point(0,0)
if M.is_O():
  print("Là gốc tọa độ")
if M.is_Ox():
  print("Nằm ở trục hoành")
if M.is_Oy():
  print("Nằm ở trục tung")
print("Khoảng cách từ gốc tọa độ: " + f"{M.distance(O)}")
```

## <font color="palered">**W8A4**</font>
[Calculator]
Xây dựng lớp Calculator (máy tính bỏ túi) với các
phương thức để thực hiện các phép toán 1. add
(cộng), 2. subtract(trừ), 3. product (nhân) và 4.
divide(chia), 5. power(hàm mũ), 6. mod (lấy phần
dư) và phương thức 7. set_numbers(cập nhật 2
số mới) nếu người dùng muốn tính toán trên 2 số
khác.
Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính,
Mỗi phương thức thực hiện một phép tính và trả
về kết quả tương ứng
Tạo menu và cho người dùng tính toán tới khi gõ
"yes", "YES" hoặc "Yes" cho câu hỏi "Bạn có
muốn thoát?" mới dừng.

```python
class Calculator:
  def __init__(self,a,b):
    self.a = a
    self.b = b
  def add(self):
    return self.a + self.b
  def subtract(self):
    return self.a - self.b
  def product(self):
    return self.a * self.b
  def divide(self):
    return self.a / self.b
  def power(self):
    return self.a ** self.b
  def mod(self):
    return self.a % self.b
  def set_numbers(self):
    x,y = map(float, input("Cập nhập số: ").split())
    self.a , self.b = x, y

running = "No"
a,b = map(float, input("Nhập số: ").split())
M = Calculator(a,b)
while running != "yes":
  print("""
  =======Máy Tính======
  1. Cộng
  2. Trừ
  3. Nhân
  4. Chia
  5. Lũy thừa
  6. Lấy dư
  7. Cập nhập số mới
  [Gõ "yes" để thoát chương trình]
  """
  )
  running = input("Nhập yêu cầu: ").lower()

  if running == "1":
    print(f"{M.a} + {M.b} = {M.add()}")
  elif running == "2":
    print(f"{M.a} - {M.b} = {M.subtract()}")
  elif running == "3":
    print(f"{M.a} x {M.b} = {M.product()}")
  elif running == "4":
    print(f"{M.a} / {M.b} = {M.divide()}")
  elif running == "5":
    print(f"{M.a} ^ {M.b} = {M.power()}")
  elif running == "6":
    print(f"{M.a} % {M.b} = {M.mod()}")
  elif running == "7":
    M.set_numbers()
print("Đã thoát!")
```
