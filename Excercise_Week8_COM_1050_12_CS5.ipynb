## <font color="palered">**W8A1**</font>
[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và
chiều cao (kiểu số thực) với hai phương thức:tính
diện tích bề mặt (diện tích toàn phần) và thể tích
của hình

**Đầu vào**

Đâu vào từ bàn phím gồm một dòng duy nhất
chứa bán kính và chiều cao của hình trụ, cách
nhau bởi một dấu cách.

**Đầu ra**

In ra màn hình diện tích bề mặt và thể tích của
hình trụ (sử dụng lại những hàm vừa viết, π=3.
14).
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
from math import pi
class Cylinder:
  def __init__(self,r,h):
    self.radius = r
    self.height = h
  def area(self):
    return (2 * (self.radius **2) * pi) + (2 * self.radius * pi * self.height )
  def  v(self):
    return  (self.radius ** 2) * pi * self.height

r, h = map(int,input().split())
n = Cylinder(r, h)
print(n.area())
print(n.v())
```

[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với
hai phương thức kiểm tra xem một ngày tháng
nào đó có hợp lệ hay không và phương thức tính
ngày kế tiếp của ngày tháng năm nào đó nếu nó là
hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm
biểu diễn dưới dạng dd/mm/yyyy). Nếu ngày
tháng năm không tồn tại, in ra dòng chữ INVALID.
Nếu là hợp lệ in ra ngày kế tiếp của nó.

**Đầu vào**
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng,
năm dưới dạng dd/mm/yyyy
.

**Đầu ra**
In ra màn hình ngày kế tiếp của ngày vừa nhập
dưới dạng dd/mm/yyyy
. Nếu ngày tháng không tồn tại, in ra INVALID
.

```python
class Date:
  def __init__(self,d,m,y):
    self.day = d
    self.month = m
    self.year = y


  def _typeYear(self):
    if (self.year % 400 == 0) or (self.year % 100 != 0 and self.year % 4 == 0):
      self.calendar = [31,29,31,30,31,30,31,31,30,31,30,31]
    else:
      self.calendar = [31,28,31,30,31,30,31,31,30,31,30,31]

  def __str__(self):
    return "What's year?"

  def is_true(self):
    if self.month < 1 or self.month > 12:
      return False
    if self.day < 1 or self.day > self.calendar[self.month - 1]:
      return False
    return True

  def next_day(self):
    self.day += 1
    if self.day > self.calendar[self.month - 1]:
      self.day = 1
      self.month += 1
      if self.month > 12:
        self.month = 1
        self.year += 1
d,m,y = map(int,input().split("/"))
n = Date(d,m,y)
n._typeYear()
if n.is_true():
  n.next_day()
  print(f"Next day: {n.day}/{n.month}/{n.year}")
else:
  print("INVALID")
```

## <font color="palered">**W8A3**</font>
[Point]
Xây dựng lớp điểm với hai thuộc tính x, y cùng
các phương thức kiểm tra xem có phải gốc tọa dộ
không, có nằm trên trục tung không, có nằm trên
trục hoành không, tính khoảng cách từ điểm đó tới
một điểm khác

**Đầu vào:**
Nhập vào tọa độ một điểm M bất kỳ

**Đầu ra:**
Thông báo M có phải gốc tọa độ không hoặc có
nằm trên trục tung hoặc trục hoành không, khoảng
cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ
số phần thập phân)

```python
import math
class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y
  def is_O(self):
    return True if (self.x == 0 and self.y == 0) else False
  def is_Ox(self):
    return True if (self.y == 0) else False
  def is_Oy(self):
    return True if (self.x == 0) else False
  def distance(self,other):
    return math.sqrt((self.x ** 2 - other.x ** 2) + (self.y ** 2 - other.y ** 2))

x,y = map(int,input("Nhập M(x,y): ").split())
M = Point(x,y)
O = Point(0,0)
if M.is_O():
  print("Là gốc tọa độ")
if M.is_Ox():
  print("Nằm ở trục hoành")
if M.is_Oy():
  print("Nằm ở trục tung")
print("Khoảng cách từ gốc tọa độ: " + f"{M.distance(O)}")
```

[Calculator]
Xây dựng lớp Calculator (máy tính bỏ túi) với các
phương thức để thực hiện các phép toán 1. add
(cộng), 2. subtract(trừ), 3. product (nhân) và 4.
divide(chia), 5. power(hàm mũ), 6. mod (lấy phần
dư) và phương thức 7. set_numbers(cập nhật 2
số mới) nếu người dùng muốn tính toán trên 2 số

Khởi tạo lớp là hai thuộc tính gồm 2 số cần tính,
Mỗi phương thức thực hiện một phép tính và trả

"yes", "YES" hoặc "Yes" cho câu hỏi "Bạn có
muốn thoát?" mới dừng.

```python
class Calculator:
  def __init__(self,a,b):
    self.a = a
    self.b = b
  def add(self):
    return self.a + self.b
  def subtract(self):
    return self.a - self.b
  def product(self):
    return self.a * self.b
  def divide(self):
    return self.a / self.b
  def power(self):
    return self.a ** self.b
  def mod(self):
    return self.a % self.b
  def set_numbers(self):
    x,y = map(float, input("Cập nhập số: ").split())
    self.a , self.b = x, y

running = "No"
a,b = map(float, input("Nhập số: ").split())
M = Calculator(a,b)
while running != "yes":
  print("""
  =======Máy Tính======
  1. Cộng
  2. Trừ
  3. Nhân
  4. Chia
  5. Lũy thừa
  6. Lấy dư
  7. Cập nhập số mới
  [Gõ "yes" để thoát chương trình]
  """
  )
  running = input("Nhập yêu cầu: ").lower()

  if running == "1":
    print(f"{M.a} + {M.b} = {M.add()}")
  elif running == "2":
    print(f"{M.a} - {M.b} = {M.subtract()}")
  elif running == "3":
    print(f"{M.a} x {M.b} = {M.product()}")
  elif running == "4":
    print(f"{M.a} / {M.b} = {M.divide()}")
  elif running == "5":
    print(f"{M.a} ^ {M.b} = {M.power()}")
  elif running == "6":
    print(f"{M.a} % {M.b} = {M.mod()}")
  elif running == "7":
    M.set_numbers()
print("Đã thoát!")
```

## <font color="palered">**W8A5**</font>
[ShoppingCart]
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý
các sản phẩm của khách. Dùng danh sách để lưu
các sản phẩm, mỗi sản phẩm là một từ điển có
name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên
để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn
từ 2 tới 7. Chương trình dừng khi người dùng
nhập "yes" cho câu hỏi "Có muốn thoát khỏi
chương trình không?"

```python
class ShoppingCart:
  def __init__(self,items = None):
    self.items = {} if items == None else items
  def add(self,name, price):
    self.items[name] = price
  def remove(self, name):
    self.items.pop(name)
  def is_None(self):
    return True if len(self.items) == 0 else False
  def total(self):
    return sum(x for x in self.items.values())
  def see_cart(self):
    for x, y in self.items.items():
      print(f"{x:<20} Giá: {y} VND")
  def remove_all(self):
    self.items.clear()

cart = ShoppingCart()
running = True
while running:
  print("""
  ==============ShoppingCart=============
  1.Thêm vào giỏ
  2.Xóa khỏi giỏ
  3.Kiểm tra giỏ
  4.Tổng tiền hiện tại
  5.Xem giỏ
  6.Xóa toàn bộ giỏ
  7.Thoát
  """
  )
  n = int(input("Nhập yêu cầu của bạn: "))
  if n == 1:
    name = input("Nhập tên món đồ: ")
    price = int(input("Nhập giá của món đồ: "))
    cart.add(name, price)
  elif n == 2:
    name = input("Nhập tên món đồ xóa: ")
    cart.remove(name)
  elif n == 3:
    print("Giỏ hàng của bạn đang trống!" if cart.is_None() else "Giỏ hàng vẫn còn đồ!")
  elif n == 4:
    print(f"Tổng thanh toán của bạn: {cart.total()} VND")
  elif n == 5:
    print("================Giỏ hàng==============")
    cart.see_cart()
  elif n == 6:
    cart.remove_all()
    print("Xóa toàn bộ!")
  elif n == 7:
    print("Bạn chắc chắn muốn thoát ?(toàn bộ giỏ hàng sẽ bị xóa):")
    n = input("Tôi chắc chắn(y/n): ")
    if n == "y":
      running = False
    else:
      pass
  print("Hẹn gặp lại")

```

thức:
area() trả về diện tích.
perimeter() trả về chu vi.
scale(k) phóng to hình chữ nhật theo hệ số k
(nhân cả w và h), không trả về gì.
Input
w h k (số thực dương)
Output
area perimeter sau khi phóng to (làm tròn 2 chữ
số thập phân).
2 5 2
10 0.5 3

```python
class Rectangle:
  def __init__(self,w,h):
    self.w = w
    self.h = h
  def area(self):
    return self.w * self.h
  def perimeter(self):
    return 2 * ( self.w + self.h )
  def scale(self,k):
    self.w = self.w * k
    self.h = self.h * k

w, h , k = map(int,input().split())
I = Rectangle(w,h)
I.scale(k)
print(f"Diện tích {I.area()}, Chu vi {I.perimeter()}")
```

## <font color="palered">**W8A7**</font>
Yêu cầu
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠
0) với:
Rút gọn phân số trong __init__ (quy ước mẫu >
0).
Các phương thức add/sub/mul/div trả về phân số
đã rút gọn.
__str__() in dạng a/b.
Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,
/})
Output
Phân số kết quả đã rút gọn
2 4 - 1 4-2 3 * 3 5
7 8 / 14 3
1 2 + 1 3
1/4-2/5
3/16
5/6

```python
from math import gcd
class Fraction:
  def __init__(self,num,den):
    self.num = num
    self.den = den
  def add(self,other):
    self.num = self.num * other.den + other.num * self.den
    self.den = self.den * other.den
  def sub(self,other):
    self.num = self.num * other.den - other.num * self.den
    self.den = self.den * other.den
  def mul(self, other):
    self.num *= other.num
    self.den *= other.den
  def div(self,other):
    self.num *= other.den
    self.den *= other.num
  def real_fraction(self):
    n = gcd(self.num,self.den)
    self.num //= n
    self.den //= n

  def __str__(self):
    return f"{self.num}/{self.den}"

n = input("Nhập yêu cầu (a b op c d): ")
a, b, op, c, d = n.split()
a, b, c, d = map(int, (a, b, c, d))

A = Fraction(a, b)
B = Fraction(c, d)
if op == "+":
  A.add(B)
elif op == "-":
  A.sub(B)
elif op == "*":
  A.mul(B)
elif op == "/":
  A.div(B)

A.real_fraction()
print(A)
```

```python
class BankAccount:
  def __init__(self,o,b):
    self.owner = o
    self.balance = b
  def deposit(self,money):
    self.balance += money
  def withdraw(self,money):
    if self.balance < money:
      print("Số dư không đủ")
    else:
      self.balance -= money
  def __str__(self):
    return f"Tài khoản {self.owner} - số dư : {self.balance}"

o, b = input("Nhập: ").split()
n = int(input("Nhập số lần: "))

b = float(b)
I = BankAccount(o,b)
while n > 0:
  f, m = input("Bạn muốn rút hay nạp tiền bao nhiêu?: ").split()
  m = float(m)
  if f.lower() == "deposit":
    I.deposit(m)
  elif f.lower() == "withdraw":
    I.withdraw(m)
  else:
    print("Vui lòng nhập đúng!")
    n += 1
  n -= 1
print(I)
```

Input

Dòng 2: n

Output
name avg rank (avg 2 chữ số).

```python
class Student:
  def __init__(self,name,scoreLst = None,scoreName = None):
    self.name = name
    self.scoreLst = [] if scoreLst is None else scoreLst
    self.scoreName = [] if scoreName is None else scoreName
  def add_score(self,subject, score):
    if subject not in self.scoreName:
      self.scoreLst.append(score)
      self.scoreName.append(subject)
  def average(self):
    return (sum(self.scoreLst) / len(self.scoreName))
  def Rank(self,avg):
    Rank = {
        5 : "Poor",
        6.5 : "Average",
        8 : "Good",
        10 : "Excellent",
    }
    for i,j in Rank.items():
      if avg < i:
        return j

name = input("Nhập tên:")
n = int(input("Số lần:"))
A = Student(name)
for i in range(n):
  a, b = input("Nhập môn / điểm : ").split()
  b = float(b)
  A.add_score(a,b)
print(f"{A.name} : {A.average():.2f} rank {A.Rank(A.average())}")


```

## <font color="palered">**W8A10**</font>
Yêu cầu
Lớp Book(title, author, year).
Lớp Library() với:
add(book)
count_by_author(name)
find_by_year(y) trả về số sách xuất bản năm y.
Input
Dòng 1: n (số thao tác)
n dòng:
ADD title;author;year
COUNT author (đếm theo tác giả)
COUNTYEAR year (đếm theo năm)
Output
Mỗi lệnh COUNT/COUNTYEAR in một dòng kết
quả.
3
ADD A;X;2000
ADD B;X;2001
COUNT X
2
4
ADD T;A;1999
ADD U;B;1999
ADD V;A;2000
COUNTYEAR 1999 2
2
ADD Only;Me;2020
COUNT You 0

```python
class Book:
  def __init__(self, title, author, year):
    self.title = title
    self.author = author
    self.year = year
  def __str__(self):
    return f"{self.title} của tác giả {self.author} xuất bản năm {self.year}"

class Library:
  def __init__(self,Storage = None):
    self.storage = [] if Storage is None else Storage
  def add(self,book):
    self.storage.append(book)
  def count_by_author(self,author):
    count = 0
    for i in self.storage:
      if i.author == author:
        count += 1
    return count
  def find_by_year(self, year):
    count = 0
    for i in self.storage:
      if i.year == year:
        count += 1
    return count

I = Library()
repeat = int(input("Nhập số lần: "))
for i in range(repeat):
  func,book = input().split()
  if func == "ADD":
    title,author,year = book.split(";")
    book = Book(title,author,year)
    I.add(book)
  elif func == "COUNT":
    print(f"{I.count_by_author(book)}")
  elif func == "COUNTYEAR":
    print(f"{I.find_by_year(book)}")
  else:
    print("Nhập sai")

```

```python
class ComplexNumber:
  def __init__(self, real, imaginary):
    self.realPart = real
    self.imaginaryPart = imaginary
  def __mul__(self,other):
    a, b = self.realPart, self.imaginaryPart
    c, d = other.realPart, other.imaginaryPart
    new_realPart = a * c - b * d
    new_imaginaryPart = a * d + b * c
    return ComplexNumber(new_realPart,new_imaginaryPart)
  def __str__(self):
    return f"{self.realPart} + {self.imaginaryPart}i"

a, b = map(float, input("Nhập số thứ nhất: ").split())
c, d = map(float, input("Nhập số thứ hai: ").split())
A = ComplexNumber(a,b)
B = ComplexNumber(c,d)
print(A * B)
```

```python
from math import sqrt
class Triangle:
  def __init__(self,Edge1,Edge2,Edge3):
    self.edge = [Edge1,Edge2,Edge3]
  def getArea(self):
    p = sum(self.edge) / 2
    return sqrt(p * (p - self.edge[0]) * (p - self.edge[1]) * (p - self.edge[2]))
  def isTriangle(self):
    if 2 * max(self.edge) < sum(self.edge):
      return True
    return False

a, b, c = map(float,input("Nhập: ").split())
T = Triangle(a,b,c)
if T.isTriangle():
  print(f"{T.getArea():.2f}")
else:
  print(-1)
```
