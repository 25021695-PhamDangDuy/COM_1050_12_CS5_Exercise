## <font color="palered">**W9A1**</font>
Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ
thuật toán sắp xếp nào

**Đầu vào**

Đầu vào từ bàn phím gồm hai dòng, mỗi dòng chứa các phần tử của một danh sách đã sắp xếp, cách nhau bởi dấu cách.

**Đầu ra**

In ra màn hình danh sách đã merge và được sắp xếp.

```python
def merge_lst(a,b):
  i = 0
  j = 0
  result = []

  while i < len(a)  and j < len(b):
    if a[i] <= b[j]:
      result.append(a[i])
      i += 1
    else:
      result.append(b[j])
      j += 1

  if i == len(a):
    result += b[j:]
  elif j == len(b):
    result += a[i:]
  return result

a = list(map(int, input("Dãy 1: ").split()))
b = list(map(int, input("Dãy 2: ").split()))
print(merge_lst(a,b))
```
